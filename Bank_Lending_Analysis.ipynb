```python
# ============================================================
# LTV CONSISTENCY VALIDATION
# ============================================================

import numpy as np

print("=" * 70)
print("LTV CONSISTENCY VALIDATION")
print("=" * 70)


# ------------------------------------------------------------
# 1. Calculate LTV from loan amount and property value
# ------------------------------------------------------------

df['calculated_ltv'] = np.where(
    (df['loan_amount'] > 0) &
    (df['property_value_clean'] > 0),
    (df['loan_amount'] / df['property_value_clean']) * 100,
    np.nan
)


# ------------------------------------------------------------
# 2. Calculate difference between reported and calculated LTV
# ------------------------------------------------------------

df['ltv_difference'] = (
    df['loan_to_value_ratio_clean'] -
    df['calculated_ltv']
)


# ------------------------------------------------------------
# 3. Count records where both LTV values are available
# ------------------------------------------------------------

valid_ltv_comparison = df[
    df['loan_to_value_ratio_clean'].notna() &
    df['calculated_ltv'].notna()
].copy()

print("\nVALID LTV COMPARISONS")
print("Number of records:", len(valid_ltv_comparison))


# ------------------------------------------------------------
# 4. Basic statistics of the difference
# ------------------------------------------------------------

print("\nLTV DIFFERENCE STATISTICS")

print(
    valid_ltv_comparison['ltv_difference'].describe()
)


# ------------------------------------------------------------
# 5. Show records with largest LTV differences
# ------------------------------------------------------------

print("\nLARGEST LTV DIFFERENCES")

print(
    valid_ltv_comparison[
        [
            'loan_amount',
            'property_value_clean',
            'loan_to_value_ratio_clean',
            'calculated_ltv',
            'ltv_difference',
            'derived_loan_product_type',
            'derived_dwelling_category',
            'total_units_clean',
            'loan_purpose',
            'action_taken'
        ]
    ]
    .sort_values(
        'ltv_difference',
        key=lambda x: x.abs(),
        ascending=False
    )
    .head(20)
    .to_string(index=False)
)


# ------------------------------------------------------------
# 6. Count records with large LTV differences
# ------------------------------------------------------------

for threshold in [1, 5, 10, 25, 50, 100]:

    count = (
        valid_ltv_comparison['ltv_difference']
        .abs() > threshold
    ).sum()

    percentage = (
        count / len(valid_ltv_comparison) * 100
    )

    print(
        f"\nAbsolute LTV difference > {threshold}: "
        f"{count} records ({percentage:.2f}%)"
    )


# ------------------------------------------------------------
# 7. Compare extreme reported LTV with calculated LTV
# ------------------------------------------------------------

print("\nREPORTED LTV ABOVE 500")

extreme_ltv_comparison = valid_ltv_comparison[
    valid_ltv_comparison['loan_to_value_ratio_clean'] > 500
]

print(
    extreme_ltv_comparison[
        [
            'loan_amount',
            'property_value_clean',
            'loan_to_value_ratio_clean',
            'calculated_ltv',
            'ltv_difference',
            'derived_loan_product_type',
            'derived_dwelling_category',
            'total_units_clean',
            'action_taken'
        ]
    ]
    .sort_values(
        'loan_to_value_ratio_clean',
        ascending=False
    )
    .head(20)
    .to_string(index=False)
)
```

```python

# ============================================================
# LTV DATA QUALITY FLAGS
# ============================================================

print("=" * 70)
print("LTV DATA QUALITY FLAGS")
print("=" * 70)


# ------------------------------------------------------------
# 1. Flag extremely high reported LTV
# ------------------------------------------------------------

df['extreme_ltv_flag'] = (
    df['loan_to_value_ratio_clean'] > 500
)


# ------------------------------------------------------------
# 2. Flag extreme disagreement between reported and
#    approximately calculated LTV
# ------------------------------------------------------------

df['ltv_consistency_flag'] = (
    df['ltv_difference'].abs() > 100
)


# ------------------------------------------------------------
# 3. Summary
# ------------------------------------------------------------

print("\nEXTREME LTV FLAG")

print(
    df['extreme_ltv_flag']
    .value_counts()
)


print("\nLTV CONSISTENCY FLAG")

print(
    df['ltv_consistency_flag']
    .value_counts()
)


# ------------------------------------------------------------
# 4. Both flags
# ------------------------------------------------------------

both_flags = df[
    df['extreme_ltv_flag'] &
    df['ltv_consistency_flag']
]

print("\nBOTH FLAGS")

print("Number of records:", len(both_flags))


print(
    "\nPercentage of dataset:",
    f"{len(both_flags) / len(df) * 100:.2f}%"
)


# ------------------------------------------------------------
# 5. Inspect flagged records
# ------------------------------------------------------------

print("\nFLAGGED RECORDS")

print(
    both_flags[
        [
            'loan_amount',
            'property_value_clean',
            'loan_to_value_ratio_clean',
            'calculated_ltv',
            'ltv_difference',
            'derived_loan_product_type',
            'derived_dwelling_category',
            'total_units_clean',
            'loan_purpose',
            'action_taken'
        ]
    ]
    .sort_values(
        'loan_to_value_ratio_clean',
        ascending=False
    )
    .head(20)
    .to_string(index=False)
)
```
